```python
import pandas as pd
import os



files = [f for f in os.listdir()
         if f.endswith(".xlsx")
        and 'common_dropped' not in f] # that not condition is important to avoid interference .otherwise it gives blank results
print(files)
```

```python
model = ['Chrononet_G', 'Chrononet_N', 'CNN_lSTM_G', 'CNN_LSTM_N', 'IC_RNN_G', 'IC_RNN_N'] # forgot '' gave error . its string
```

```python
desired_acc= 80
```

```python
common_dropped = None
```

```python


all_data = []
common_dropped = None

for file in files:
    df = pd.read_excel(file)

    
    # clean dropped column first
    df["dropped"] = df["dropped"].astype(str).str.strip() # to remove space / any thing blank 
    df = df.nlargest(50, "val_acc")
    all_data.append(df)
    

    dropped_values = set(df["dropped"])

    if common_dropped is None:
        common_dropped = dropped_values
    else:
        common_dropped &= dropped_values


print("Common dropped:", common_dropped)


# Collect val_acc from every file
result_list = []

for i, df in enumerate(all_data):
    temp = df[
        (df["dropped"].isin(common_dropped)) #&
       # (df["val_acc"] > desired_acc)
    ][
        ["dropped", "val_acc"]
    ].copy()

    temp["model"] = f"{model[i]}"
    result_list.append(temp)


result = pd.concat(result_list, ignore_index=True)

result = result.sort_values(
    by="val_acc",
    ascending=False
)

print(result)

result.to_excel("common_dropped_all_files.xlsx", index=False)
```
