```python
import os
import pandas as pd 
```

# extracting channel names and val_acc



```python
# channel names 
files = []
dropped_channels =[]
directory = r"F:\WORK\Thesis\code_reducing_channel\outputs\guinea_outputs\outputs_CNN_LSTM\keep12"

for filename in os.listdir(directory):
     #print(filename)
     files.append(filename)

     channel = filename[-7:]
     if channel[0] == '_':
        channel = channel[1:]
     if channel[:2]=='p_':
        channel= channel[2:]

     dropped_channels.append(channel)
     print(channel)
     
print(f'{len(files)} files found and {len(dropped_channels)} dropped channels  found')     
```

```python
#val acc
c=0
prev_csv = []
for root, dirs, files in os.walk(directory):
    for filename in files:
        if filename.endswith(".csv"):
            print(os.path.join(root, filename))
            c+=1
            prev_csv.append(os.path.join(root,filename))
print(f'{c} csv files found')            
```

```python

val_acc  = [] 
for i in prev_csv:
    df  = pd.read_csv(i)
    val_acc.append(df['val_accuracy'].max())
    print(df['val_accuracy'].max())
len(val_acc)    

```
